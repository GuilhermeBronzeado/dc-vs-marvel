# 02 — Limpeza e filtro Marvel/DC

Lê:    `data/raw/total_comics_1997_2019.csv`
Grava: `data/processed/marvel_dc_1997_2019.csv`

Objetivo: manter só Marvel e DC, preservando o Rank original entre TODAS as editoras.

```python
import os
# Sobe para a raiz do projeto (dc-vs-marvel) caso o kernel esteja em notebooks/
if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")
print("cwd:", os.getcwd())
```

```python
import pandas as pd

df = pd.read_csv("data/raw/total_comics_1997_2019.csv")
df.shape
```

```python
df.dtypes
df.isna().sum()
df["Editora"].value_counts().head(10)
```

```python
value_counts = df["Editora"].value_counts()
value_counts[value_counts.index.str.contains("marvel|dc", case=False)]
```

```python
df["Editora"] = df["Editora"].str.strip()

df["Editora"].value_counts().head(3)
```

```python
marvel_dc = df[df["Editora"].isin(["Marvel", "DC"])].copy()
len(marvel_dc), len(marvel_dc) / len(df)
```

```python
marvel_dc["Editora"].value_counts()
```

```python
marvel_dc.groupby(["Ano", "Editora"]).size().unstack(fill_value=0)
```

```python
marvel_dc.to_csv("data/processed/marvel_dc_1997_2019.csv", index=False)
print("salvou", len(marvel_dc), "linhas")
```

## Resultado

- Dataset filtrado: 45.841 linhas (Marvel 22.443 + DC 23.398)
- Período: 1997-01 a 2019-12 (276 meses)
- Colunas: Ano, Mes, Rank, Titulo, Edicao, Preco, Editora
- Rank preservado entre todas as editoras originais (não é rank só de Marvel+DC)
- Arquivo: `data/processed/marvel_dc_1997_2019.csv`
