# 01 — Coleta de dados

Fonte: ComicChron (monthlycomicssales)
Período: 1997-01 a 2019-12
Saída: `data/raw/total_comics_1997_2019.csv`

As funções de parsing estão em `src/coleta.py`.

```python
import sys
sys.path.append("..")

import pandas as pd
import time
from src.coletaComics import extrair_mes
```

```python
lista_dfs = []
ano, mes = 1997, 1
fim = (2019, 12)

while (ano, mes) <= fim:

    print(f"{ano}-{mes:02d}", end=" ", flush=True)

    df_mes = extrair_mes(ano, mes)

    if df_mes is not None and len(df_mes) > 0:
        lista_dfs.append(df_mes)
        
        n = len(df_mes)
        if n < 90:
            marca = "  <-- baixo"
        else:
            marca = ""
        print(f"({n}){marca}")

    else:
        print("(falhou)")

    time.sleep(1)
    mes += 1

    if mes > 12:
        mes = 1
        ano += 1
```

```python
df = pd.concat(lista_dfs, ignore_index=True)
df.to_csv("data/raw/total_comics_1997_2019.csv", index=False)
print(f"Total: {len(df)} registros")
print(df.groupby("Ano").size())
```

## Resultado

- 90.833 registros em 276 meses
- 0 falhas, 0 duplicatas
- Arquivo salvo em `data/raw/total_comics_1997_2019.csv`
